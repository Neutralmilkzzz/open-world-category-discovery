```python
import numpy as np
def clustering_accuracy(y_true, y_pred):
    n_clusters = np.max(y_pred) + 1
    n_classes = np.max(y_true) + 1

    M = np.zeros((n_clusters, n_classes), dtype=int)

    # TODO: Count samples for every (cluster, class) pair
    # n = len(y_true)
    # i = 0
    # while (i < n):
        # M[y_pred[i]][y_true[i]] += 1
        # i += 1
    for pred, true in zip(y_pred, y_true):
        M[pred, true] += 1


    from scipy.optimize import linear_sum_assignment

    row_ind, col_ind = linear_sum_assignment(M, maximize=True)

    print("Rows:", row_ind)
    print("Columns:", col_ind)
    print("Matched counts:", M[row_ind, col_ind])

    acc = M[row_ind, col_ind].sum() / M.sum()
    return acc
```

```python

# Test 1: Permuted cluster IDs
print(clustering_accuracy(
    np.array([0, 0, 1, 1, 2, 2]),
    np.array([2, 2, 0, 0, 1, 1])
))
# Expected: 1.0

# Test 2: One class split into two clusters
print(clustering_accuracy(
    np.array([0, 0, 0, 0, 1, 1]),
    np.array([0, 0, 2, 2, 1, 1])
))
# Expected: 0.666666...
```

```python

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt

X, y_true = make_blobs(
    n_samples=600,
    centers=4,
    cluster_std=1.2,
    random_state=42
)

# Simulate labeled data from classes 0 and 1 only
labeled_mask = (y_true == 0) | (y_true == 1)

print("All samples:", len(X))
print("Labeled samples:", labeled_mask.sum())
print("Known classes:", np.unique(y_true[labeled_mask]))

pairs = []
for k in range(2, 9):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    y_pred = model.fit_predict(X) #本质就是 反复执行assignment和update直到收敛
    print(f"Clustering accuracy for k={k}: {clustering_accuracy(y_true[labeled_mask], y_pred[labeled_mask]):.4f}")
    pairs.append((k,clustering_accuracy(y_true[labeled_mask], y_pred[labeled_mask])))
```
